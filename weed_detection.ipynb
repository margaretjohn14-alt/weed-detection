2. Training

````python
from ultralytics import YOLO

model = YOLO('yolov8n.pt')

results = model.train(
    data='./Weeds-3/data.yaml',
    epochs=20,
    imgsz=640,
    batch=16,
    name='weed_detection',
    project='runs'
)
````

3. Evaluation & Metrics

````python
from ultralytics import YOLO
import glob, os

model = YOLO('/content/runs/detect/runs/weed_detection-4/weights/best.pt')
test_imgs = glob.glob('Weeds-3/test/images/*.jpg')[:10]

os.makedirs('inference_examples', exist_ok=True)
model.predict(source=test_imgs, save=True, project='inference_examples', name='results', conf=0.35)
````
